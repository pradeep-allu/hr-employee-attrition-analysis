a)Descriptive Statistics

```python
for col in numerical_cols:
    print("col_name =",col,"\n",df[col].describe())
    print("====================")
```

b) Numerical Univariate Analysis — Distribution Analysis

```python
plot_number = 1

plt.figure(figsize= (15,10))
for col in numerical_cols:
    plt.subplot(3,5,plot_number)
    sns.histplot(x = df[col])
    plt.xticks(rotation = 45)
    plot_number += 1
plt.tight_layout() 
```

```python
plot_number = 1

plt.figure(figsize= (15,10))
for col in numerical_cols:
    plt.subplot(3,5,plot_number)
    sns.boxplot(x = df[col])
    plt.xticks(rotation = 45)
    plot_number += 1
plt.tight_layout() 
```

```python
cat_vs_attrition = [
    'businesstravel',
    'department',
    'educationfield',
    'gender',
    'jobrole',
    'maritalstatus',
    'overtime'
]
```

```python
for col in cat_vs_attrition:
    print("-----",col,"vs Attrition -------")
    print(pd.crosstab(df[col],df["attrition"]))
```

```python
for col in cat_vs_attrition:
    print("----",col,"------")
    print(pd.crosstab(df[col],df["attrition"],normalize= "index")*100)
```

```python
plot_number = 1

plt.figure(figsize= (15,12))
for col in cat_vs_attrition:
    plt.subplot(3,3,plot_number)
    sns.countplot(data = df, x = col, hue = "attrition")
    plt.title(f"{col} vs Attrition")
    plt.xticks(rotation = 90)
    plot_number += 1
plt.tight_layout()
plt.show()
```
